```python
// subprocess 2 code goes here: Handle NaN Values
# === SUBPROCESS 2: HANDLE NaN VALUES (REMOTE-SENSING SAFE) ===
print("\n" + "=" * 70)
print("SUBPROCESS 2: HANDLING NaN VALUES (REMOTE-SENSING SAFE)")
print("=" * 70)

# 1. Replace infinities (from DEM math errors like divide-by-zero)
df.replace([np.inf, -np.inf], np.nan, inplace=True)

# 2. Define what NOT to impute/scale
target_col = 'SUSCEP'
spatial_cols = ['X', 'Y']
non_feature_cols = [target_col] + spatial_cols

# Define predictive feature columns
feature_cols = [c for c in df.columns if c not in non_feature_cols]

# 3. Replace GIS NoData sentinel values (< -1e30 as well as standard -9999)
for col in feature_cols:
    df.loc[df[col] < -1e30, col] = np.nan

sentinel_values = [-9999, -999]
for val in sentinel_values:
    df.replace(val, np.nan, inplace=True)

# 4. Impute ONLY features, column by column (median is robust to skew)
for col in feature_cols:
    missing_before = df[col].isnull().sum()
    if missing_before > 0:
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"  → {col}: Filled {missing_before} NaN/sentinel value(s) with median = {median_val:.4f}")

# 5. Handle target: drop rows where SUSCEP is missing (NEVER impute labels)
rows_before = len(df)
df.dropna(subset=[target_col], inplace=True)
rows_dropped = rows_before - len(df)
if rows_dropped > 0:
    print(f"  → Dropped {rows_dropped} rows with missing target '{target_col}'")
else:
    print(f"  → Target '{target_col}' has 0 missing values (all rows retained).")

# 6. Verify
assert df.isnull().sum().sum() == 0, "NaN values still remain!"
print("\n✓ All NaN values successfully handled.")
```

```python
// subprocess 3 code goes here : Standardise values

print("\n" + "=" * 70)
print("SUBPROCESS 3: STANDARDISING VALUES")
print("=" * 70)

target_col = 'SUSCEP'
spatial_cols = ['X', 'Y']

# 1. Define predictive features (numeric only, excluding target and coordinates)
feature_cols = [c for c in df.columns 
                if c not in [target_col] + spatial_cols 
                and np.issubdtype(df[c].dtype, np.number)] 

print(f"Features to standardise: {feature_cols}")
print(f"Preserving as-is: {spatial_cols + [target_col]}")

# 2. Separate features (X) and target (y)
X = df[feature_cols].copy()
y = df[target_col].copy()

# 3. Standardise ONLY the predictive features (Z-Score: mean=0, std=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Build clean output DataFrame preserving original column order and index
df_scaled = df.copy()
df_scaled[feature_cols] = X_scaled

print("\n✓ Standardisation complete using StandardScaler (Z-Score)")
print(f"\nPost-Standardisation Summary:")
print(df_scaled[feature_cols].describe().loc[['mean', 'std', 'min', 'max']].round(4))

print("\n=== STANDARDIZED DATA HEAD ===")
print(df_scaled.head())
```
